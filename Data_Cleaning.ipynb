```python
import pandas as pd
import numpy as np

print("Libraries imported successfully!")
```

```python
# Load Dataset

df = pd.read_csv("retail_sales_dataset.csv")

print("Dataset loaded successfully!")
print("Shape:", df.shape)
```

```python
# Inspect Dataset

print("Dataset Shape:")
print(df.shape)

print("\nColumn Names:")
print(df.columns.tolist())

print("\nData Types:")
print(df.dtypes)

print("\nFirst 5 Rows:")
display(df.head())
```

```python
# Check Missing Values

missing_values = df.isnull().sum()

print("Missing Values in Each Column:")
print(missing_values)
```

```python
# Check Duplicate Records

duplicate_count = df.duplicated().sum()

print("Number of Duplicate Records:", duplicate_count)
```

```python
# Convert Date Column to Datetime

df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

print("Date column converted successfully!")
print("Date Data Type:", df["Date"].dtype)
```

```python
# Check Invalid Dates

invalid_dates = df["Date"].isna().sum()

print("Number of Invalid Dates:", invalid_dates)
```

```python
# Check Outliers using IQR Method

numeric_columns = [
    "Age",
    "Quantity",
    "Price per Unit",
    "Total Amount"
]

for column in numeric_columns:
    Q1 = df[column].quantile(0.25)
    Q3 = df[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_limit = Q1 - 1.5 * IQR
    upper_limit = Q3 + 1.5 * IQR

    outliers = df[
        (df[column] < lower_limit) |
        (df[column] > upper_limit)
    ]

    print(f"{column}: {len(outliers)} outliers")
```

```python
# Final Data Cleaning Check

print("Missing Values:")
print(df.isnull().sum())

print("\nDuplicate Records:", df.duplicated().sum())

print("\nInvalid Dates:", df["Date"].isna().sum())

print("\nDataset Shape:", df.shape)
```

```python
# Save Cleaned Dataset

df.to_csv("cleaned_retail_sales_dataset.csv", index=False)

print("Cleaned dataset saved successfully!")
```

## Data Cleaning Summary

* The dataset contains 1,000 records and 9 columns.
* No missing values were found in the dataset.
* No duplicate records were found.
* The Date column was converted from string format to datetime format.
* No invalid dates were found.
* No statistical outliers were detected in the numerical columns using the IQR method.
* The cleaned dataset was saved as `cleaned_retail_sales_dataset.csv`.

### Conclusion

The dataset was successfully inspected and cleaned. It did not require removal of missing values, duplicate records, or outliers. The cleaned dataset is ready for further analysis.
